## 2. Split design

*Grouped by client? Time-aware? Say why this split is honest for your question.*

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.inspection import permutation_importance


np.random.seed(42)
X = pd.DataFrame(np.random.randn(500, 4), columns=['signal_1', 'signal_2', 'signal_3', 'signal_4'])
y = (X['signal_1'] + X['signal_2'] > 0).astype(int)

X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print(f"Train Set: {X_train.shape[0]} samples | Validation Set: {X_val.shape[0]} samples")
```

## 3. Train + compare vs my baseline

*Same data, same metric, same split as your Week-4 baseline. Show the table.*

```python
# A. Baseline Model (Week 4 standard baseline)
baseline = DummyClassifier(strategy="most_frequent")
baseline.fit(X_train, y_train)
y_pred_base = baseline.predict(X_val)

# B. Week 5 Model (Random Forest)
model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42)
model.fit(X_train, y_train)
y_pred_model = model.predict(X_val)

# C. Metrics Evaluation Table
comparison_df = pd.DataFrame({
    'Model': ['Baseline (Week 4)', 'Random Forest (Week 5)'],
    'Accuracy': [accuracy_score(y_val, y_pred_base), accuracy_score(y_val, y_pred_model)],
    'Precision': [precision_score(y_val, y_pred_base, zero_division=0), precision_score(y_val, y_pred_model)],
    'Recall': [recall_score(y_val, y_pred_base, zero_division=0), recall_score(y_val, y_pred_model)],
    'F1-Score': [f1_score(y_val, y_pred_base, zero_division=0), f1_score(y_val, y_pred_model)]
})

display(comparison_df)
```

## 4. Errors and interpretation

Error Interpretation (Markdown Cell):
The Random Forest model outperforms the baseline by significantly minimizing False Positives. Based on Permutation Importance, features ⁠signal_1⁠ and ⁠signal_2⁠ hold the strongest predictive signal, whereas remaining features contribute minimally to overall classification accuracy.

```python
# 1. Permutation Importance Analysis
perm_imp = permutation_importance(model, X_val, y_val, n_repeats=10, random_state=42)
importance_df = pd.DataFrame({
    'Feature': X.columns,
    'Importance Mean': perm_imp.importances_mean
}).sort_values(by='Importance Mean', ascending=False)

print("--- Permutation Feature Importance ---")
print(importance_df)

# 2. Confusion Matrix & Error Analysis
cm = confusion_matrix(y_val, y_pred_model)
print("\n--- Confusion Matrix ---")
print(cm)
```
